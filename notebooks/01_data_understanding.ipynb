# Data Understanding

We will work with a **bank churn database**, which maintains the following fields:

| Field | Description |
|---|---|
| **RowNumber** | The row number of the record in the dataset |
| **CustomerId** | A unique ID number for each bank customer |
| **CreditScore** | A number that shows how good the customer is with credit |
| **Age** | The age of the customer in years |
| **Tenure** | Number of years the customer has been with the bank |
| **Balance** | The amount of money currently in the customer's account |
| **NumOfProducts** | How many bank products (like accounts or loans) the customer uses |
| **HasCrCard** | Shows if the customer has a credit card (1 = Yes, 0 = No) |
| **IsActiveMember** | Shows if the customer uses the bank frequently (1 = Yes, 0 = No) |
| **EstimatedSalary** | The estimated amount of money the customer earns |
| **Exited** | Shows if the customer left the bank (1 = Left, 0 = Stayed) |
| **Complain** | Shows if the customer made a complaint (1 = Yes, 0 = No) |
| **Satisfaction Score** | The score given by the customer for solving their complaint |
| **Point Earned** | Points accumulated by the customer for using bank services |
