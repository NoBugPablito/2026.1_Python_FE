```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("../database/Customer-Churn-Records.csv")
print(df.columns)

```

## 1. Remove Irrelevant Variables

`RowNumber`, `CustomerId`, and `Surname` were removed because they mainly function as identifiers and do not provide predictive value for the model.

```python
df = df.drop( columns=['RowNumber', 'CustomerId', 'Surname'])

print(df.columns)

df.to_csv("../DataBase/data_cleaning/data_cleaning.csv", index=False, encoding="utf-8")
```

## 2. Missing values and duplicates 

No missing values or duplicated records were detected in the dataset, therefore no imputation or duplicate removal procedures were required.

# 3. Skewness

Based on the EDA results, variables such as `Age` and `NumOfProducts` presented moderate skewness and some extreme observations. However, these values appear plausible within the banking context and may contain relevant predictive information.

For this reason, no outlier removal or aggressive transformations were applied during the cleaning stage.

## 4. Categorical Variables 

Based on the previous categorical analysis, variables such as `Geography`, `Gender`, and `Card Type`were retained due to their potential predictive relevance for customer churn modeling.

These variables will be transformed during the preprocessing stage using dummy variables to measure their behavior.

## 5. The correlation analysis

The correlation analysis performed during the EDA stage revealed that `Complain` has an extremely strong relationship with the target variable `Exited`.

Although this relationship may be reasonable from a business perspective, it could also indicate potential target leakage. Therefore, two modeling approaches will be evaluated:
- one including `Complain`,
- and another excluding it,

in order to compare model behavior and predictive performance.
